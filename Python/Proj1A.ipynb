```python
import pandas as pd
df = pd.read_csv("RAW_Data.csv")

# Count actual missing values in each column
print(df.isna().sum())

#Calculate missing percentages:
missing_percent = (df.isna().mean() * 100).round(2)
print(missing_percent)

#Calculate missing percentages:
missing = df.isna().sum()
print(missing[missing > 0])

#Check text placeholders:
placeholders = ["N/A", "unknown", "-"]
for column in df.select_dtypes(include="object").columns:
    count = df[column].astype("string").str.strip().str.lower().isin(
        [x.lower() for x in placeholders]
    ).sum()

    print(column, count)
```

```python
import pandas as pd
df = pd.read_csv("RAW_Data.csv")

#Find exact duplicate rows
#This counts duplicate rows, excluding the first occurrence of each repeated row.
print(df.duplicated().sum())

#To display them:
#keep=False displays all rows belonging to duplicate groups.
duplicates = df[df.duplicated(keep=False)]
print(duplicates)

#Check duplicate transaction IDs
print(df["Transaction_Line_ID"].duplicated().sum())

#Check duplicate order IDs
print(df["Order_ID"].duplicated().sum())

#Find repeated identifiers
repeated_ids = df[
    df.duplicated(subset=["Transaction_Line_ID"], keep=False)
]

print(repeated_ids)

#Count distinct orders
print(df["Order_ID"].nunique())
```

```python
import pandas as pd
df = pd.read_csv("RAW_Data.csv")

#nspect all data types
print(df.dtypes)

#Get a detailed overview
df.info()

#Inspect date values
print(df["Order_Date"].head(20))

#Parse dates in a separate column
df["Order_Date_Parsed"] = pd.to_datetime(
    df["Order_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

#Identify values that failed parsing
invalid_dates = df[
    df["Order_Date"].notna() &
    df["Order_Date_Parsed"].isna()
]

print(invalid_dates[["Order_Date"]])

#Investigate numeric fields
print(df["Quantity"].value_counts(dropna=False))
#To inspect suspicious values:
print(df["Quantity"].unique()[:30])
```

```python
import pandas as pd
df = pd.read_csv("RAW_Data.csv")

#Check minimum and maximum
print(df[["Quantity", "Unit_Price", "Unit_Cost",
          "Sales", "Profit"]].describe())

#Identify negative values
print(df[df["Quantity"] < 0])
#For prices:
print(df[df["Unit_Price"] < 0])
#For profit:
print(df[df["Profit"] < 0])
#Find zero values
print(df[df["Quantity"] == 0])

#Identify potential outliers using IQR
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[
    (df["Sales"] < lower) |
    (df["Sales"] > upper)
]
print(outliers)
```
