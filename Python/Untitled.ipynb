```python
#bar chart
products = ["Laptop", "Phone", "Tablet", "Watch"]
revenue = [50000, 80000, 30000, 20000]

plt.figure(figsize=(8, 4))
plt.bar(products, revenue)
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (₹)")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#line chart
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
monthly_sales = [20000, 24000, 22000, 30000, 35000, 40000] 
plt.figure(figsize=(8, 4))
plt.plot(
    months,
    monthly_sales,
    marker="o",
    label="Sales"
)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
