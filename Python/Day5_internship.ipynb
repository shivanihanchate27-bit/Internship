```python
#calculating sum
import numpy as np
sales = np.array([100, 200, 300, 400, 500])
total = np.sum(sales)
print(total)

#The mean is the arithmetic average.
average = np.mean(sales)
print(average)                                                                         

#Returns the smallest value.
print(np.min(sales))                                                               

#Returns the largest value.
print(np.max(sales))                                                              
```
