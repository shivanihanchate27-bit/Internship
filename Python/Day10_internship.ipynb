```python
import numpy as np
import pandas as pd
from collections import Counter

sales = [100, 120, 120, 130, 140, 150, 160, 180, 500]

df = pd.DataFrame({"Sales": sales})

print(df)

mean_value = df["Sales"].mean()
print("Mean:", mean_value)

median_value = df["Sales"].median()
print("Median:", median_value)

mode_values = df["Sales"].mode()
print("Mode:", mode_values.tolist())

std_dev = df["Sales"].std(ddof=0)
print("Population standard deviation:", std_dev)

data = pd.Series([10, 20, 30, 40, 50, 60, 70, 80])

print("Q1:", data.quantile(0.25))
print("Q2:", data.quantile(0.50))
print("Q3:", data.quantile(0.75))

p90 = df["Sales"].quantile(0.90)
print("90th percentile:", p90)

#interquartile
q1 = df["Sales"].quantile(0.25)
q3 = df["Sales"].quantile(0.75)

iqr = q3 - q1
print("IQR:", iqr)
```
