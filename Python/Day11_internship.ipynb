```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Revenue": [12000, 15000, 14000, 18000, 22000, 25000],
    "Profit": [2000, 2800, 2400, 3500, 4500, 5200],
    "Orders": [120, 140, 135, 160, 190, 210],
    "Advertising": [1000, 1200, 1100, 1500, 1800, 2000]
}

df = pd.DataFrame(data)

print(df)

fig, ax = plt.subplots(figsize=(8, 5))

#bar chart
plt.figure(figsize=(8, 5))

plt.bar(df["Month"], df["Revenue"], label="Revenue")

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()

#line chart
plt.figure(figsize=(8, 5))

plt.plot(
    df["Month"],
    df["Revenue"],
    marker="o",
    label="Revenue"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

#histogram
order_values = [
    100, 120, 130, 140, 150,
    160, 170, 180, 200, 220,
    250, 280, 300, 350, 500
]
plt.figure(figsize=(8, 5))

plt.hist(order_values, bins=6, edgecolor="black")

plt.title("Distribution of Order Values")
plt.xlabel("Order Value (₹)")
plt.ylabel("Number of Orders")

plt.tight_layout()
plt.show()

#scatter plot
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Advertising",
    y="Revenue"
)

plt.title("Advertising vs Revenue")
plt.xlabel("Advertising Expense (₹)")
plt.ylabel("Revenue (₹)")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
