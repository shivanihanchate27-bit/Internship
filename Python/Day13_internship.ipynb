```python
import pandas as pd

data = {
    "OrderID": [
        "O101", "O102", "O103", "O104", "O105",
        "O106", "O107", "O108", "O109", "O110"
    ],
    "Date": [
        "2026-01-05", "2026-01-12", "2026-01-20",
        "2026-02-03", "2026-02-14", "2026-02-22",
        "2026-03-02", "2026-03-10", "2026-03-18",
        "2026-03-25"
    ],
    "Product": [
        "Laptop", "Phone", "Laptop", "Headphones", "Phone",
        "Laptop", "Phone", "Headphones", "Laptop", "Phone"
    ],
    "Region": [
        "West", "East", "West", "North", "West",
        "South", "East", "West", "North", "West"
    ],
    "Quantity": [2, 3, 1, 5, 2, 1, 4, 10, 2, 1],
    "UnitPrice": [50000, 20000, 50000, 3000, 20000,
                  50000, 20000, 3000, 50000, 20000],
    "UnitCost": [40000, 15000, 40000, 1800, 15000,
                 40000, 15000, 1800, 40000, 15000]
}

df = pd.DataFrame(data)
df["Date"] = pd.to_datetime(df["Date"])

df["Sales"] = df["Quantity"] * df["UnitPrice"]
df["Cost"] = df["Quantity"] * df["UnitCost"]
df["Profit"] = df["Sales"] - df["Cost"]

print(df.head())

#Question 1 — Which products generate the most sales and profit?
product_summary = (
    df.groupby("Product")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Units_Sold=("Quantity", "sum")
      )
      .sort_values("Total_Sales", ascending=False)
)

print(product_summary)

#Question 2 — Which region performs best?
region_summary = (
    df.groupby("Region")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum")
      )
)

region_summary["Profit_Margin"] = (
    region_summary["Total_Profit"]
    / region_summary["Total_Sales"] * 100
)

print(region_summary.sort_values("Total_Sales", ascending=False))

#Question 3 — How do sales change month by month?
monthly_summary = (
    df.groupby(df["Date"].dt.to_period("M"))
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum")
      )
)

print(monthly_summary)

#Question 4 — Which products have the highest profit margin?
margin_summary = (
    df.groupby("Product")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum")
      )
)

margin_summary["Profit_Margin"] = (
    margin_summary["Total_Profit"]
    / margin_summary["Total_Sales"] * 100
)

print(margin_summary.sort_values("Profit_Margin", ascending=False))

#Question 5 — Are there data-quality problems?

# 1. Dataset size
print("Rows and columns:", df.shape)

# 2. Missing values
print("\nMissing values:")
print(df.isnull().sum())

# 3. Duplicate records
print("\nDuplicate rows:", df.duplicated().sum())

# 4. Duplicate order IDs
print("Duplicate Order IDs:", df["OrderID"].duplicated().sum())

# 5. Invalid quantities or prices
print("Invalid quantities:", (df["Quantity"] <= 0).sum())
print("Invalid prices:", (df["UnitPrice"] < 0).sum())
print("Invalid costs:", (df["UnitCost"] < 0).sum())

# 6. Date range
print("Date range:", df["Date"].min(), "to", df["Date"].max())

# 7. Check whether cost exceeds price
print("Rows where cost exceeds price:",
      (df["UnitCost"] > df["UnitPrice"]).sum())
```
