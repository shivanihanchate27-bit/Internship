```python
#import csv file
import csv
with open("sales.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Product", "Quantity", "Price"])
    writer.writerow(["Pen", 10, 5])
    writer.writerow(["Notebook", 4, 40])
    writer.writerow(["Folder", 3, 20])
print("CSV file created.")
```

```python
import csv
import os

filename = "students.csv"

try:
    # Check whether the file exists
    if not os.path.exists(filename):
        print("Error: CSV file not found.")

    else:
        # Open and read the CSV file
        with open(filename, "r", newline="") as file:
            reader = csv.DictReader(file)

            for row in reader:
                try:
                    name = row["Name"]
                    marks = float(row["Marks"])

                    if not name.strip() or not 0 <= marks <= 100:
                        raise ValueError("Invalid student data")

                    print(name, marks)

                except ValueError:
                    print("Invalid data found:", row)

except (OSError, csv.Error) as error:
    print("Error:", error)
```
